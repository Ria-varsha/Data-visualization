```python
df = pd.read_csv("tourism_exact_190_raw.csv")
```

```python
df = df.dropna(subset=["Country", "ISO", "Continent"])
df = df.fillna(0)
```

```python
print(df.columns)
```

```python
df["Arrivals_2019"] = pd.to_numeric(df["Arrivals_2019"], errors="coerce")
df["Arrivals_2023"] = pd.to_numeric(df["Arrivals_2023"], errors="coerce")
```

```python
df = df[df["Arrivals_2019"] > 0]
```

```python
df["pct_growth"] = ((df["Arrivals_2023"] - df["Arrivals_2019"]) / df["Arrivals_2019"]) * 100
```

```python
df["pct_growth"] = df["pct_growth"].round(1)
```

```python
df = df.drop_duplicates(subset=["ISO"])
```

```python
df["Country"] = df["Country"].str.strip()
df["Continent"] = df["Continent"].str.strip()
df["ISO"] = df["ISO"].str.strip().str.upper()
```

```python
df_clean = df[["Country", "ISO", "Continent", "pct_growth"]]
```

```python
df_clean.head()
df_clean.shape
df_clean.describe()
```

```python
df_clean.to_csv("tourism_cleaned.csv", index=False)
```
